````python
import torch
import numpy as np
````

First, we create a set_seed() function that sets the random seed for both NumPy and PyTorch. Using the same seed allows us to reproduce the same random values each time the code is executed.

````python

def set_seed(seed):
    np.random.seed(seed)
    torch.manual_seed(seed)
````

Using a seed of 42, we generate six random integers from 0 to 4 and store them in a NumPy array. The resulting array is [3, 4, 2, 4, 4, 1], which confirms that the seed produces the expected values.

````python
set_seed(42)
arr = np.random.randint(0, 5, size=6).astype('int32')

assert np.array_equal(arr, np.array([3,4,2,4,4,1])), "Oops"
print(arr)
````

We then convert the NumPy array into a PyTorch tensor using torch.from_numpy(). This allows us to continue working with the data using PyTorch operations.

````python
x = torch.from_numpy(arr)
````

The tensor initially uses a 32-bit integer type. We change it to 64-bit integer, or int64, which PyTorch represents as a LongTensor. The output confirms the change from `torch.IntTensor` to `torch.LongTensor`.

````python
print('Old:', x.type())

x = x.type(torch.int64)

print('New:', x.type())
````

Next, we reshape the tensor from a one-dimensional structure into a 3-by-2 tensor. This gives us a matrix that can be accessed by rows and columns.

````python
x = x.reshape(3,2)
````

From the reshaped tensor, we extract the right-hand column using column indexing. The resulting values are `[4, 4, 1]`.

````python
print(x[:,1:])
````

We then demonstrate an element-wise operation by squaring the values of x. The important point here is that `x.pow(2)` creates the squared values without changing the original tensor. The original tensor remains `[[3, 4], [2, 4], [4, 1]]`.

````python
x.pow(2)
print(x)
````

For the matrix multiplication, we create another PyTorch tensor called `y`. Since `x` has a shape of 3 by 2, `y` needs to have a compatible shape of 2 by 3. Using the same seed of 42 produces the matrix `[[2, 2, 1], [4, 1, 0]]`.

````python
# Code here
set_seed(42)
y = torch.randint(0,5,(2,3))

assert torch.equal(y, torch.tensor([[2, 2, 1],[4, 1, 0]])), "Oops"
print(y)
````

Finally, we multiply x and y using PyTorch's matrix multiplication function. Since x is 3 by 2 and y is 2 by 3, the resulting matrix is 3 by 3. The product is `[[22, 10, 3]`, `[20, 8, 2]`, `[12, 9, 4]]`.

````python
prod = torch.mm(x,y)
prod
````
