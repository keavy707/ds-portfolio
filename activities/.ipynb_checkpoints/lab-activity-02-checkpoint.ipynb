## **Forward Propagation and Error Computation Using ReLU Activation**

**Forward propagation** (or a forward pass) is the process by which an artificial neural network processes input data to generate a prediction. During this process, data flows in a single direction—starting from the input layer, moving through any hidden layers, and finishing at the output layer. <br>

___

For this laboratory activity, we are performing forward propagation and error computation using the ReLU activation function. We will use a small two-layer neural network and manually follow how the input passes through the hidden layer and output layer until we get the final prediction and error.

$$\text{hidden unit weights} = 
\begin{bmatrix}
w_{11} = 0.2 && w_{12} = -0.3 \\
w_{13} = 0.4 && w_{14} = 0.1 \\
w_{15} = -0.5 && w_{16} = 0.2
\end{bmatrix}$$

$$\text{output unit weights} = 
\begin{bmatrix}
w_{21} = -0.3 \\
w_{22} = -0.2
\end{bmatrix}$$

$$\theta = 
\begin{bmatrix}
\theta_{1} = -0.4 \\
\theta_{2} = 0.2 \\
\theta_{3} = 0.1
\end{bmatrix}$$
</div>

The process follows four steps:

1. **Weighted sums (hidden layer):** Each hidden unit computes a weighted sum of the inputs plus a bias term (Z = W·x + θ).
2. **Activation (ReLU):** Each Z value passes through f(Z) = max(0, Z), which zeroes out any negative values and leaves positive values unchanged.
3. **Output layer:** The hidden activations are combined through the output weights and bias to produce a final weighted sum, which also passes through ReLU to produce the prediction ŷ.
4. **Error:** The difference between the prediction and the true label, δ = ŷ − y, quantifies how far off the network's guess was.


**Setup and Initialization**<br>

```python
import numpy as np
```

```python
x = np.array([1, 0, 1])  # Input shape: (3,)
y = np.array([1])        # Target shape: (1,)


# Activation Function: ReLU
def relu(z):
    return np.maximum(0, z)

# Weights and Biases (Thetas)
# Hidden unit weights: 3 inputs -> 2 hidden units
W_h = np.array([
    [ 0.2, -0.3],  # w11, w12
    [ 0.4,  0.1],  # w13, w14
    [-0.5,  0.2]   # w15, w16
])

# Output unit weights: 2 hidden units -> 1 output unit
W_o = np.array([
    [-0.3],        # w21
    [-0.2]         # w22
])

# Biases
theta_1_2 = np.array([-0.4, 0.2]) 
theta_3 = np.array([0.1])         
```

**Forward Pass:**

```python
# --- Forward Pass ---

# 1. Hidden Layer
Z_h = np.dot(x, W_h) + theta_1_2   
H = relu(Z_h)                       

# 2. Output Layer
Z_o = np.dot(H, W_o) + theta_3    
y_hat = relu(Z_o)                 

# 3. Compute Error (y_hat - y)
error = y_hat - y
```

**Step-by-Step Manual Calculation**

Hidden Layer Weighted Sums ($Z_h$):
- $Z_1 = (1 \cdot 0.2) + (0 \cdot 0.4) + (1 \cdot -0.5) - 0.4 = -0.7$
- $Z_2 = (1 \cdot -0.3) + (0 \cdot 0.1) + (1 \cdot 0.2) + 0.2 = 0.1$

Hidden Layer Activations ($H = \max(0, Z)$):
- $H_1 = \max(0, -0.7) = 0$
- $H_2 = \max(0, 0.1) = 0.1$

Output Layer Weighted Sum $Z_3$) & Prediction ($\hat{y}$):
- $Z_3 = (0 \cdot -0.3) + (0.1 \cdot -0.2) + 0.1 = 0.08$
- $\hat{y} = \max(0, 0.08) = 0.08$

Error Calculation:
- $\text{Error} = \hat{y} - y = 0.08 - 1 = -0.92$

```python
# Display Results
print(f"Z (Hidden Sums)   : {Z_h}")
print(f"H (Hidden Outputs): {H}")
print(f"Z_o (Output Sum)  : {Z_o}")
print(f"y_hat (Prediction): {y_hat[0]}")
print(f"Error (y_hat - y) : {error[0]}")
```

>A key detail worth noting: because ReLU zeroes out negative inputs, the first hidden unit (Z₁ = −0.7) is deactivated (H₁ = 0). Therefore, it does not contribute to the final output. 


Overall, the forward pass shows how the input is progressively transformed through the network. The ReLU activation deactivates the first hidden unit because its weighted sum is negative, leaving only the second hidden unit to contribute to the final prediction of \(0.08\). Since this is below the target value of \(1\), the resulting error is \(-0.92\).
